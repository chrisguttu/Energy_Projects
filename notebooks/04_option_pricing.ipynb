```python
# tell the notebook where to look
import sys
import os
# Add the src folder to the system path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..", "src")))

# Import models and packages
from Option_models.black_76 import Black76
from Option_models.implied_vol import implied_vol
from Option_models.bachelier import Bachelier
```

### Define some input variables

```python
# create some variables 
F = 105.0       # Forward price
K = 100.0       # Strike price
T = 0.5         # Time to maturity (6 months)
r = 0.02        # Risk-free rate (2%)
true_sigma = 0.30  # True volatility (30%)
```

## Black 76


```python
# Test Black 76
model = Black76(F=F, K=K, T=T, r=r, sigma=true_sigma)
market_price = model.call_price()
print(f"Market price (from true sigma): {market_price:.4f}")

```

```python
# Get the Implied Volatility of the option price
market_price = 11.5
iv = implied_vol(
    price=market_price,
    model_class=Black76,
    option_class="Call",
    F=F,
    K=K,
    T=T,
    r=r
)

print(f"Implied volatility: {iv:.4f}")
```

```python
# Test Bachelier

sigma_n = true_sigma * F

model = Bachelier(F=F, K=K, T=T, r=r, sigma_n=sigma_n)
market_price = model.call_price()
print(f"Market price (from true sigma): {market_price:.4f}")

model.greeks(option_type="Call")
```
